```python
from rdkit import Chem

mol = Chem.MolFromSmiles('CCO')  # Ethanol
print(mol.GetNumAtoms())          # 3
```

```python
from rdkit import Chem
from rdkit.Chem import Descriptors #molecule ki properties 

# Aspirin
mol = Chem.MolFromSmiles('CC(=O)Oc1ccccc1C(=O)O')

print("Molecular Weight:", Descriptors.MolWt(mol))
print("Number of Atoms:", mol.GetNumAtoms())
print("Number of Rings:", mol.GetRingInfo().NumRings())
```

```python
# Tumhara apna compound!
mol = Chem.MolFromSmiles('C1=CC=CC=C1C(C[Se]C(C2=CC=CC=C2)=O)=O')
print("Molecular Weight:", Descriptors.MolWt(mol))
print("Number of Atoms:", mol.GetNumAtoms())
print("Number of Rings:", mol.GetRingInfo().NumRings())
```

```python
from rdkit import Chem
from rdkit.Chem import Descriptors

mol = Chem.MolFromSmiles('C1=CC=CC=C1C(C[Se]C(C2=CC=CC=C2)=O)=O')

mw = Descriptors.MolWt(mol)
logp = Descriptors.MolLogP(mol)
hbd = Descriptors.NumHDonors(mol)
hba = Descriptors.NumHAcceptors(mol)

print("Molecular Weight:", mw)
print("LogP:", logp)
print("H-Bond Donors:", hbd)
print("H-Bond Acceptors:", hba)

# Lipinski Rule Check
if mw <= 500 and logp <= 5 and hbd <= 5 and hba <= 10:
    print("✅ Drug-like molecule!")
else:
    print("❌ Not drug-like")
```
